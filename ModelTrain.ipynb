# Iris Dataset - Model Training
Train a Random Forest classifier on the Iris dataset and evaluate it on a separate test set.

```python
# Import required libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

## Load training data

```python
# Load the training dataset
df = pd.read_csv('Iris_try.csv')
df
```

```python
# Split into features (x) and target (y)
x = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

## Train the model

```python
# Create and train a Random Forest classifier
model = RandomForestClassifier()
model.fit(x, y)
```

## Load test data and predict

```python
# Load the test dataset
df2 = pd.read_csv("Iris_test.csv")
```

```python
# Features for the test set
x_test = df2.iloc[:, :-1]
```

```python
# Actual labels for the test set
y_test = df2.iloc[:, -1]
actual = y_test.values
```

```python
# Predict species for the test set
prediction = model.predict(x_test.values)

print(prediction)
```

## Check accuracy

```python
# Compare predictions with actual labels to calculate accuracy
count = 0
for i in range(0, len(actual)):
    if actual[i] == prediction[i]:
        count += 1
accuracy = (count * 100) / len(prediction)
print(f"The accuracy is {accuracy}")
```
